## 1.1 Functions and Function Composition

A neural network is essentially a massive, nested mathematical function. Before we can optimize it, we must understand how basic functions connect.

### Mathematical Definition
A function \(f\) maps an input \(x\) to a single output \(y\):
\[y = f(x)\]

When we pass the output of one function into another, we create a **composite function**:
\[z = g(f(x))\]

> **Why this matters for LLMs:** Every layer in a deep learning model is a function. A 12-layer Transformer network is just a function composition written as \(f_{12}(f_{11}(...f_1(x)...))\).

### Python Implementation
Here is how we represent function composition cleanly in Python:

```python
def f(x):
    return x ** 2  # Squaring function

def g(x):
    return x + 3   # Adding a bias

# Compute g(f(x)) where x = 5
input_val = 5
result = g(f(input_val))
print(f"Result of g(f({input_val})): {result}")  # Outputs 28
```
